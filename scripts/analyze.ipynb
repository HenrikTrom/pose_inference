```python
import onnx
import onnxruntime as ort

# Load the ONNX model
model_path = '/home/docker/workspace/build/dependencies/Docker-OpenCV-TensorRT-Dev/modules/pose_inference/models/end2end.onnx'  # replace with your actual model path
onnx_model = onnx.load(model_path)

# Check that the model is well-formed
onnx.checker.check_model(onnx_model)
print("ONNX model loaded and checked.")

# Set up an ONNX Runtime session
session = ort.InferenceSession(model_path)

# Get output details
outputs = session.get_outputs()

# Print output names and shapes
print("Model outputs:")
for output in outputs:
    print(f"Name: {output.name}, Shape: {output.shape}, Type: {output.type}")
```
